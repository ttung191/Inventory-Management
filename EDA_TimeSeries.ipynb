```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.statespace.sarimax import SARIMAX
import warnings
warnings.filterwarnings("ignore")
```

```python
conn = sqlite3.connect('data/inventory.db')
df = pd.read_sql_query("SELECT * FROM Nhat_Ky_Giao_Dich WHERE Loai_GD='Xuat' AND Ma_Hang='CAT_VANG'", conn)
df['Ngay_Giao_Dich'] = pd.to_datetime(df['Ngay_Giao_Dich'])
df.set_index('Ngay_Giao_Dich', inplace=True)
```

```python
print("--- KẾT QUẢ KIỂM ĐỊNH ADF ---")
monthly_sales = df.resample('M').size()
result = adfuller(monthly_sales)
print(f'ADF Statistic: {result[0]}')
print(f'p-value: {result[1]}')
if result[1] <= 0.05:
    print("=> THÀNH CÔNG: p-value <= 0.05, bác bỏ H0. Dữ liệu ĐÃ DỪNG (Stationary), đủ điều kiện chạy SARIMA.")
else:
    print("=> THẤT BẠI: Dữ liệu chưa dừng, cần lấy sai phân (Differencing).")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 4))
plot_acf(monthly_sales, ax=axes[0], lags=20)
plot_pacf(monthly_sales, ax=axes[1], lags=20)
plt.show()
```

```python
monthly_sales = df.resample('M').size()
model = SARIMAX(monthly_sales, order=(1, 0, 1), seasonal_order=(1, 0, 1, 12))
results = model.fit()
print(results.summary())
results = model.fit()
print(results.summary())
```
